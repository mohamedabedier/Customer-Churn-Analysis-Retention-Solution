# IQR for Detecting Outiers

```python
def IQR(df, column):
    q1, q3 = df.approxQuantile(column, [0.25, 0.75], 0.01)
    median = df.approxQuantile(column, [0.5], 0.01)[0]

    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    print(f"{column}")
    print(f"Q1     : {q1}")
    print(f"Q3     : {q3}")
    print(f"Median : {median}")
    print(f"IQR    : {iqr}")
    print(f"Lower  : {lower}")
    print(f"Upper  : {upper}")

    return median
```

# Dealing with N/A

```python
# Preserve accepted as nullable boolean, add explicit tri-state string for warehouse loading

df_offers = df_offers.withColumn(
    "accepted_flag",
    F.when(F.col("accepted") == True, "Yes")
     .when(F.col("accepted") == False, "No")
     .otherwise("Unknown")
)
```

```python
# Impute missing date_offered with sentinel date for warehouse date-dimension join

df_offers = df_offers.withColumn(
    "date_offered_key",
    F.coalesce(F.col("date_offered"), F.to_date(F.lit("1900-01-01")))
)
```

### Offers — Missing Value Imputation

Before loading into the warehouse, missing values in `offer_type`, `accepted`, and `date_offered` were handled as follows. Rows were **not dropped** for missing dimensional attributes — only `offer_id` nulls would justify dropping a row, since that breaks the fact table's grain.

**`offer_type` (78 nulls)**
- Imputed with `"Unknown"` — a low missing rate on a categorical attribute, safe to fill with an explicit placeholder rather than lose the row.

**`accepted` (152 nulls)**
- Left as a nullable boolean (no True/False was guessed, to avoid fabricating signal in the acceptance rate).
- Added `accepted_flag` (`"Yes"` / `"No"` / `"Unknown"`) as an explicit tri-state column for warehouse loading, so NULLs are never silently coerced to False downstream.

**`date_offered` (64 nulls)**
- No statistical imputation (mean/mode) was used, since there's no meaningful "average" offer date and a fabricated date would corrupt recency-based features.
- Added `date_offered_key`, coalescing nulls to a sentinel date (`1900-01-01`), so the row can still join cleanly to a `Dim_Date` table without breaking on NULL foreign keys.
- The original `date_offered` column is preserved (still nullable) for feature engineering steps (e.g. "days since last offer"), where sentinel/null rows should be filtered out explicitly rather than treated as real dates.

## Spark SQL — Monthly Churn Rate 

```python
churn_by_cohort = (df_customers
    .filter(F.col("date_opened").isNotNull() & F.col("is_churned").isNotNull())
    .withColumn("cohort_month", F.date_format("date_opened", "yyyy-MM"))
    .groupBy("cohort_month")
    .agg(
        F.count("*").alias("total_customers"),
        F.sum(F.col("is_churned").cast("int")).alias("churned_customers"))
    .withColumn("churn_rate_pct", F.round(F.col("churned_customers") / F.col("total_customers") * 100, 2))
    .orderBy("cohort_month"))

churn_by_cohort.show(30)
```

### Interpretation — Cohort Churn Rate by Signup Month

**Target of this calculation**: for each signup month (`date_opened`), what percentage of
customers who opened their account that month are *currently* marked as churned
(`is_churned = 1`). This is a **cohort snapshot**, not a time-series churn rate — it doesn't
show churn *happening* in a given month, only the churn status *today* of everyone who joined
in that month. True monthly churn (customers who left *during* that specific month) would need
a churn-event date, which doesn't exist in this data.

**What the output presents**: one row per signup month, with `total_customers` in that cohort,
how many of them are `churned_customers`, and the resulting `churn_rate_pct`.

# 🏆 MODEL LAYER — Churn Prediction (SVM)
Reads `Dim_Customer` from Hive and trains a binary classifier for `is_churned`.

```python
from pyspark.sql import SparkSession, functions as F
from pyspark.sql.window import Window
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler, StandardScaler
from pyspark.ml.classification import LinearSVC
from pyspark.ml import Pipeline
from pyspark.ml.evaluation import BinaryClassificationEvaluator, MulticlassClassificationEvaluator

spark = (SparkSession.builder
    .appName("Churn_Model_QuickRun")
    .master("local[*]")
    .config("spark.sql.session.timeZone", "UTC")
    .getOrCreate())
spark.sparkContext.setLogLevel("WARN")
```

```python
COLS = ["cust_key", "customer_id", "age", "gender", "geography", "credit_score", "balance",
        "tenure_months", "num_products", "has_cr_card", "is_active", "estimated_salary",
        "is_churned", "CLV_LTV", "avg_ticket_res_time_hrs", "dw_start_date", "dw_end_date"]

raw = spark.read.option("header", False).csv("hdfs:///churn_proj/dim_customer_csv/Dim_Customer.csv")
raw = raw.toDF(*COLS)

dim = (raw
    .withColumn("cust_key", F.col("cust_key").cast("int"))
    .withColumn("customer_id", F.col("customer_id").cast("long"))
    .withColumn("age", F.col("age").cast("int"))
    .withColumn("credit_score", F.col("credit_score").cast("int"))
    .withColumn("balance", F.col("balance").cast("double"))
    .withColumn("tenure_months", F.col("tenure_months").cast("int"))
    .withColumn("num_products", F.col("num_products").cast("int"))
    .withColumn("has_cr_card", F.col("has_cr_card").cast("boolean"))
    .withColumn("is_active", F.col("is_active").cast("boolean"))
    .withColumn("estimated_salary", F.col("estimated_salary").cast("double"))
    .withColumn("is_churned", F.col("is_churned").cast("boolean"))
    .withColumn("CLV_LTV", F.col("CLV_LTV").cast("double"))
    .withColumn("avg_ticket_res_time_hrs", F.col("avg_ticket_res_time_hrs").cast("double")))

dim.printSchema()
dim.show(5)
print("Rows loaded:", dim.count())
```

```python
print("--- Null counts ---")
dim.select([F.sum(F.col(c).isNull().cast("int")).alias(c) for c in dim.columns]).show()

print("--- is_active vs is_churned (leakage check) ---")
dim.groupBy("is_active", "is_churned").count().orderBy("is_active", "is_churned").show()

print("--- Class balance ---")
dim.groupBy("is_churned").count().show()
churn_rate = dim.selectExpr("avg(cast(is_churned as double)) as rate").first()["rate"]
print(f"Overall churn rate: {churn_rate:.4f}")
```

```python
INCLUDE_IS_ACTIVE = False

dim_ml = (dim
    .fillna({"avg_ticket_res_time_hrs": 0.0, "CLV_LTV": 0.0})
    .withColumn("label", F.col("is_churned").cast("double"))
    .withColumn("has_cr_card_int", F.col("has_cr_card").cast("int"))
    .dropna(subset=["age", "credit_score", "balance", "num_products", "tenure_months",
                     "estimated_salary", "gender", "geography", "label"]))

print("Rows going into modeling:", dim_ml.count())

NUMERIC_COLS = ["age", "credit_score", "balance", "num_products", "tenure_months",
                "estimated_salary", "CLV_LTV", "avg_ticket_res_time_hrs", "has_cr_card_int"]
if INCLUDE_IS_ACTIVE:
    NUMERIC_COLS.append("is_active_int")
    dim_ml = dim_ml.withColumn("is_active_int", F.col("is_active").cast("int"))

CATEGORICAL_COLS = ["gender", "geography"]

indexers = [StringIndexer(inputCol=c, outputCol=f"{c}_idx", handleInvalid="keep") for c in CATEGORICAL_COLS]
encoder = OneHotEncoder(inputCols=[f"{c}_idx" for c in CATEGORICAL_COLS],
                         outputCols=[f"{c}_ohe" for c in CATEGORICAL_COLS])
assembler = VectorAssembler(inputCols=NUMERIC_COLS + [f"{c}_ohe" for c in CATEGORICAL_COLS],
                             outputCol="features_raw")
scaler = StandardScaler(inputCol="features_raw", outputCol="features", withMean=True, withStd=True)
```

```python
train_df, test_df = dim_ml.randomSplit([0.8, 0.2], seed=42)
print("Train rows:", train_df.count(), " Test rows:", test_df.count())
```

```python
train_churn_rate = train_df.selectExpr("avg(label)").first()[0]
pos_weight = (1 - train_churn_rate) / train_churn_rate if 0 < train_churn_rate < 1 else 1.0
train_df_w = train_df.withColumn("class_weight", F.when(F.col("label") == 1, F.lit(pos_weight)).otherwise(F.lit(1.0)))
print(f"Train churn rate: {train_churn_rate:.4f} -> positive class weight: {pos_weight:.3f}")

svm = LinearSVC(featuresCol="features", labelCol="label", weightCol="class_weight", maxIter=50, regParam=0.1)
pipeline = Pipeline(stages=indexers + [encoder, assembler, scaler, svm])

model = pipeline.fit(train_df_w)
predictions = model.transform(test_df)
predictions.select("customer_id", "label", "prediction", "rawPrediction").show(10)
```

```python
auc_eval = BinaryClassificationEvaluator(labelCol="label", rawPredictionCol="rawPrediction", metricName="areaUnderROC")
acc_eval = MulticlassClassificationEvaluator(labelCol="label", predictionCol="prediction", metricName="accuracy")
f1_eval  = MulticlassClassificationEvaluator(labelCol="label", predictionCol="prediction", metricName="f1")

print(f"AUC:      {auc_eval.evaluate(predictions):.4f}")
print(f"Accuracy: {acc_eval.evaluate(predictions):.4f}")
print(f"F1:       {f1_eval.evaluate(predictions):.4f}")

predictions.groupBy("label", "prediction").count().orderBy("label", "prediction").show()
```

```python
model.write().overwrite().save("hdfs:///churn_proj/models/svm_churn_pipeline")
print("Model saved.")
```
